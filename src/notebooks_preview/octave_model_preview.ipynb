```python
from jupyterscad import view

from tmkb.configuration import load_config
from tmkb.kb_modules import OctaveBuilder, BackplateBuilder

from pathlib import Path
```

```python
conf = load_config(Path("../configs/default.conf.yaml"))
conf
```

# Octave model

The main model of the keyboard - here all the keys sits. To emulate piano feel black keys are a bit higher than white one.

```python
builder = OctaveBuilder(7, conf)
backplate = BackplateBuilder(builder, conf)
backplate.move([0, backplate.dy + 10, 0])
view(builder.build() + backplate.build())
```
