```python
from jupyterscad import view

from pathlib import Path

from tmkb.parts.controller_case import CrontrollerCaseBuilder
from tmkb.kb_modules.octave import OctaveBuilder
from tmkb.configuration import load_config
from tmkb.parts.backplate import BackplateBuilder
```

```python
conf = load_config(Path("../configs/default.conf.yaml"))
conf
```

```python
builder = OctaveBuilder(7, conf)
```

```python
case = CrontrollerCaseBuilder(builder, conf)

backplate = BackplateBuilder(case, conf)
backplate.move([0, 0, -5])
view(case.build() + backplate.build())
```

```python
conf.output_dir.mkdir(parents=True, exist_ok=True)
case.build().save_as_stl(conf.output_dir / "case_v1.stl")
```
