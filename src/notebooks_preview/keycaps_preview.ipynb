```python
from jupyterscad import view

from tmkb.configuration import load_config
from tmkb.kb_modules import MxKeyBuilder, MxKeyType
from pathlib import Path
```

```python
conf = load_config(Path("../configs/default.conf.yaml"))
```

# Black Key

Half tone keys, used in upper half of the keyboard

```python
black_key = MxKeyBuilder.mx_key_factory(MxKeyType.BLACK, conf)
view(black_key.build())
```

# White key

Full tones, used in lower half of the keyboard.

```python
white_key = MxKeyBuilder.mx_key_factory(MxKeyType.WHITE, conf)
view(white_key.build())
```

# Octave controller keys

Used as `octave up/down` controllers at the microcontroller case

```python
octave_key = MxKeyBuilder.mx_key_factory(MxKeyType.OCTAVE_CONTROLLER, conf)
view(octave_key.build())
```
