```python
from jupyterscad import view

from tmkb.configuration import load_config
from tmkb.kb_modules import MxKeyBuilder
from pathlib import Path
```

```python
conf = load_config(Path("../configs/default.conf.yaml"))
```

```python
key = MxKeyBuilder.mx_key_factory(True, conf)
```

```python
view(key.build())
```

```python
conf.output_dir.mkdir(parents=True, exist_ok=True)

key.to_stl(conf.output_dir / "white_keycap.stl")
```
